# Deep-Learning-Finance-Volume-1 — end-to-end demo

This notebook walks through the full pipeline: load prices, engineer features,
build sliding-window sequences, train the `DirectionLSTM`, and backtest a
simple long/flat strategy. It uses reproducible synthetic data so it runs
offline.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from dlfinance.data import load_prices, PriceDataset
from dlfinance.model import DirectionLSTM
from dlfinance import __version__

print('dlfinance version:', __version__)
```

```python
prices = load_prices(source='synthetic', seed=7)
data = PriceDataset.from_prices(prices, window=20)
print('prices:', prices.shape[0])
print('train/val/test:', len(data.x_train), len(data.x_val), len(data.x_test))
print('features:', data.n_features)
```

```python
from argparse import Namespace
from dlfinance.train import train

args = Namespace(
    source='synthetic', ticker='AAPL', period='5y', window=20,
    epochs=40, batch_size=64, hidden_size=32, lr=1e-3, seed=7,
    output_dir='../outputs',
)
metrics = train(args)
metrics
```

The `training_report.png` written to `../outputs/` shows the training curve
and the backtest of the LSTM long/flat strategy versus buy-and-hold.

To try real market data instead, set `args.source = 'yfinance'` and
`args.ticker = 'MSFT'` (requires network access).
